### Q1. Write a program to calculate the sum of first n terms of a geometric progression with scale factor a and common ratio r. (Example of input and output: Input: n = 4, a = 2, r = 3. Output will be sum of 2 + 6 + 18 + 54 = 80)

```python
n = eval(input("Pass the iteration value n:"))
a = eval(input("Pass the value a:"))
r = eval(input("Pass the value r:"))

sum1=0
list1 = []

for i in list(range(1,n+1)):
    #print(i)
    if i == 1:
        sum1 = a + sum1*r
        
    else:
         sum1 = sum1*r
            
    list1.append(sum1)
    print(list1)

final_sum = sum(list1)
print("Final Result : ", final_sum)
```
