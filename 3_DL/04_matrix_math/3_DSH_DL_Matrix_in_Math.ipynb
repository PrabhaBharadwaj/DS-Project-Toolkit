<h3 style='color:blue'>Calculate total sales from units and price per unit using matrix multiplication</h3>

```python
price_per_unit = np.array([1000,400,1200])
units = np.array([[30,40,50],[5,10,15],[2,5,7]])
```

```python
price_per_unit
# 1*3
```

```python
units
# 3*3
```

```python

price_per_unit*units  # this is simple multiplication where in price_per_unit gets multiplied with all rows of units.


# In normal multiplication it converts price_per_unit = np.array([1000,400,1200],[1000,400,1200],[1000,400,1200])
```

In above case numpy is using broadcasting so it expands price_per_unit array from 1 row, 3 columns to 3 row and 3 columns. Correct way to do matrix multiplication is to use dot product as shown below

### DOT PRODUCT

```python
np.dot(price_per_unit,units)  
# 1*3  *  3*3  = 1*3

# This is dot product where in row gets multiplied at column level i.e price_per_unit gets multiplied with first column of units and gets added.
# Then again price_per_unit gets multiplied with 2nd column of units and gets added. 
```
