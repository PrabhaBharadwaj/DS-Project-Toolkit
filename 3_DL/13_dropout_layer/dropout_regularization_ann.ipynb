<h2 align='center' style='color:blue'>Dropout Regularization In Deep Neural Network</h2>

- In Deep learning, to avoid model over-fitting, we uses Dropout Regularization method.
- Randomly dropping few neurons/Hidden units from overfit models hidden layer, this method is called Dropout Regularization.

This is a dataset that describes sonar chirp returns bouncing off different services. The 60 input variables are the strength of the returns at different angles. It is a **binary classification problem** that requires a model to differentiate rocks from metal cylinders.

- Here Output has category as metal and rocks

```python
X = df.drop(60, axis=1)
y = df[60]
y.head()
```

## Train test split

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=1)
```

```python
X_train.head()
```

# Using Deep Learning Model

## Model without Dropout Layer

```python
import tensorflow as tf
from tensorflow import keras
```

```python
# Here 60,30,15 ouput dim are randomly selected,last 1 is the actual based on output req. 
#Here op is binary result, so chose 1 hidden unit as o/p

model = keras.Sequential([
    keras.layers.Dense(60, input_dim=60, activation='relu'),
    keras.layers.Dense(30, activation='relu'),
    keras.layers.Dense(15, activation='relu'),
    keras.layers.Dense(1, activation='sigmoid')
])

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
# as its binary classification problem we have used 'binary_crossentropy'

model.fit(X_train, y_train, epochs=100, batch_size=8)  # Here used Mini batch Gradient Descent with batch size=8
```

```python
model.evaluate(X_test, y_test)
```

> Here accuracy of train data  =1 , but with test data its 0.7692 --> So this looks like overfit model

Training Accuracy >>> Test Accuracy

```python
y_pred = model.predict(X_test).reshape(-1)
print(y_pred[:10])

# round the values to nearest integer ie 0 or 1
y_pred = np.round(y_pred)
print(y_pred[:10])
```

```python
y_test[:10]
```

```python
from sklearn.metrics import confusion_matrix , classification_report

print(classification_report(y_test, y_pred))
```

### Model with Dropout Layer
- Here dropout applied after each layer, here 0.5 means 50% hiddenunits are dropped randomly

```python
modeld = keras.Sequential([
    keras.layers.Dense(60, input_dim=60, activation='relu'),
    keras.layers.Dropout(0.5), 
    keras.layers.Dense(30, activation='relu'),
    keras.layers.Dropout(0.5),
    keras.layers.Dense(15, activation='relu'),
    keras.layers.Dropout(0.5),
    keras.layers.Dense(1, activation='sigmoid')
])

modeld.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

modeld.fit(X_train, y_train, epochs=100, batch_size=8)
```

```python
modeld.evaluate(X_test, y_test)
```

> Here accuracy of train data  = 0.8269 , and with test data its 0.7692 --> So this looks like NOT overfit model, both accuracy are almost nearer

- So this dropout helped to reduce overfit problem. This mainly used in computer vision, where complex neural network with high overfit problem


Training Accuracy is still good but Test Accuracy Improved

```python
y_pred = modeld.predict(X_test).reshape(-1)
print(y_pred[:10])

# round the values to nearest integer ie 0 or 1
y_pred = np.round(y_pred)
print(y_pred[:10])
```

```python
from sklearn.metrics import confusion_matrix , classification_report

print(classification_report(y_test, y_pred))
```

**You can see that by using dropout layer test accuracy increased from 0.75 to 0.77**
- Here dropout applied randomly only random hidden units, so accuracy will be different in each run
