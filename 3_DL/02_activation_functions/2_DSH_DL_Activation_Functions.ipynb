<h3 style='color:purple'>Sigmoid</h3>  Value varies between 0 to 1

- Output between 0 to 1. Output will be maximum likelihood, it will be probabilistic for each class, so max one will be considered later

```python
import math

def sigmoid(x):
  return 1 / (1 + math.exp(-x))
```

```python
sigmoid(100)
```

```python
sigmoid(1)
```

```python
sigmoid(-56)   # below value is e to the power -25, so its very nearer to 0. 
```

```python
sigmoid(0.5)
```

<h3 style='color:purple'>tanh</h3> Values varies between -1 and +1

```python
def tanh(x):
  return (math.exp(x) - math.exp(-x)) / (math.exp(x) + math.exp(-x))
```

```python
tanh(-56)
```

```python
tanh(-2)
```

```python
tanh(-0.2)
```

```python
tanh(50)
```

```python
tanh(1)
```

<h3 style='color:purple'>ReLU</h3> Value varies between 0 to +value


- If Value<0 then O/P=0 If Value >0 then O/P= value

```python
def relu(x):
    return max(0,x)
```

```python
relu(-100)
```

```python
relu(8)
```

```python
relu(800)
```

<h3 style='color:purple'>Leaky ReLU</h3>

- If Value<0 then O/P~0 , close to 0. If Value >0 then O/P= value

```python
def leaky_relu(x):
    return max(0.1*x,x)
```

```python
leaky_relu(-100)   # max(0.1*-100, -100)
```

```python
leaky_relu(8)
```

```python
leaky_relu(800)
```
