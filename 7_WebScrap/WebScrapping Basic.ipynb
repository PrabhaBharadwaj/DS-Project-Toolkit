# READ UPPINAKUDRU DATA

```python
res1 = requests.get('https://en.wikipedia.org/wiki/Uppinakudru')
```

```python
#soup1 = bs4.BeautifulSoup(res1.text, 'lxml').get_text(strip=True)
soup1 = bs4.BeautifulSoup(res1.text, 'lxml')
#  Here .get_text(strip=True) will remove extra character inbetween the text 
```

```python
soup1
# Search some character for table data to read only that
```

```python
l1 = []
for i in soup1.select('.infobox-label') :
    print(i.text)
    j = i.text
    j = j.replace(u'\xa0', u' ') # Here it removes extra padding '\xa0•\
    l1.append(j)
    
print(l1)
```

```python
l2 =[]
for i in soup1.select('.infobox-data') :
    print(i.text)
    j = i.text
    j = j.replace(u'\xa0', u' ') # Here it removes extra padding '\xa0•\
    l2.append(j)
    
print(l2)
```

```python
l1 = pd.Series(l1)
l2 = pd.Series(l2)
```

```python
l1 = pd.DataFrame(l1, columns=['Detail'])
l2 = pd.DataFrame(l2, columns=['Detail2'])
l1
```

```python
l3 = pd.concat([l1,l2],axis=1)

l3
```
