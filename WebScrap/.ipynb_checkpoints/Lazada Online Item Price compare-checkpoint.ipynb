- Link: https://www.lazada.sg/catalog/?_keyori=ss&from=input&q=mask

```python
# import library
from bs4 import BeautifulSoup
import requests
import pandas as pd

# Request to website and download HTML contents
#url='https://www.lazada.sg/catalog/?_keyori=ss&from=input&q=mask'
url='https://www.lazada.com.ph/shop-fashion-face-mask/'
req=requests.get(url)
content=req.text
print(content)
```

- I used requests library to get data from website. You can see that so far what we have is unstructured text.

##### Step 3: Format the downloaded content into readable format

```python
soup=BeautifulSoup(content)
# This step is very straight-forward and what we do is just parse unstructured text into Beautiful Soup 
# and what you get is as below.
```

```python
soup
```

> The output is a much more readable format and you can search different HTML elements or class in it.

##### Step 4: Extract out useful information and save into a structured format


> This step requires some time to understand website structure and find out where the data is stored exactly. For the Lazada case, it is stored in a Script section as JSON format.

```python
raw=soup.findAll('script')[3].text
page=pd.read_json(raw.split("window.pageData=")[1],orient='records')
```

```python
#raw=soup.findAll('script')[1]
```

```python
#Store data
for item in page.loc['listItems','mods']:
    brand_name.append(item['brandName'])
    price.append(item['price'])
    location.append(item['location'])
    description.append(ifnull(item['description'],0))
    rating_score.append(ifnull(item['ratingScore'],0))
```
