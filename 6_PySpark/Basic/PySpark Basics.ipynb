## Drop the NAN Value

```python
df_pyspark.na.drop().show()
# Drop parameter : drop(how='any', thresh=None, subset=None)
#Wher-ever null is there, that whole row will be deleted
# Here not assigned to df_pyspark, so not changed in DF
```

```python
df_pyspark.show()
```

```python
### how==any This is default how , If any value in null then it removes whole row
df_pyspark.na.drop(how='any').show()
```

```python
### how==all , If all value in null then it removes whole row
df_pyspark.na.drop(how='all').show()
```

```python
### thresh=None is default: Thresh means minimum those many colu value in row is not null then it wont drops

df_pyspark.na.drop(thresh=1).show()
```

```python
### thresh=2 :  means minimum 2  colu value in row is not null then it keeps that row, else drops, in last row only 1 is not null, so dropped

df_pyspark.na.drop(thresh=2).show()
```

```python
df_pyspark.na.drop(thresh=3).show()
```

```python
### subset==None This is default value , If we want to check only specific columns null value and delete entire row, then we keep subset

df_pyspark.na.drop(how='any',subset=['Experience'] ).show()
```

```python
df_pyspark.na.drop(how='any',subset=['age'] ).show()
```

### MISSING VALUE: Fill the missing/NAN value with Mean value like Imputation in  panda

```python
from pyspark.ml.feature import Imputer

imputer = Imputer(
                inputCols = ['age','Experience','Salary'],
                outputCols = ["{}_imputed".format(c) for c in  ['age','Experience','Salary']] # Here it creates new col as _imputed
                ).setStrategy("mean")

# setStrategy  we can change as mean, mode, median etc
```

```python
# Add Imputation cols to df
imputer.fit(df_pyspark).transform(df_pyspark).show()
```

```python
from pyspark.ml.feature import Imputer

imputer = Imputer(
                inputCols = ['age','Experience','Salary'],
                outputCols = ["{}_imputed".format(c) for c in  ['age','Experience','Salary']] # Here it creates new col as _imputed
                ).setStrategy("median")

# setStrategy  we can change as mean, mode, median etc
# Add Imputation cols to df
imputer.fit(df_pyspark).transform(df_pyspark).show()
```

### FILTER OPERATION: BY PYSPARK

```python
#### SALARY of people less than equal to 30000
# Way1
df_pyspark.filter("Salary<=30000").show()
```

```python
# Selected column
df_pyspark.filter("Salary<=30000").select(['name','age']).show()
```

```python
# Way2
df_pyspark.filter(df_pyspark['Salary']<=30000).select(['name','age']).show()
```

```python
# Way3
## Muktiple cond
df_pyspark.filter((df_pyspark['Salary']<=30000) & 
                  (df_pyspark['Salary']>=15000 )).select(['name','age','Salary']).show()
```

```python
## Not operation ~
df_pyspark.filter(~(df_pyspark['Salary']<=30000)).select(['name','age','Salary']).show()
```

### Group by and Aggregate

- Group by AND Aggreagation comes together
    
- Different aggregation Functions are : agg, avg,sum,count,max,means,pivot etc

```python
df_pyspark1 = spark.read.option('header','true').csv('test2.csv', inferSchema='true')
```

```python
df_pyspark1.show()
```

```python
df_pyspark1.printSchema()
```

```python
## Group by

df_pyspark1.groupBy('name')
```

```python
df_pyspark1.groupBy('name').sum()


# Here "DataFrame[name: string, sum(Salary): bigint]"  means group by on 'name' and aggregation/Sum function on "salary".
# Always aggregation/Sum happens on integer value, not on string
```

```python
## Grouped to find the total sum salary of each person
df_pyspark1.groupBy('name').sum().show()
```

```python
## Grouped to find the max salary of a person
df_pyspark1.groupBy('name').max().show()
```

```python
## Grouped to find the max salary of a person
df_pyspark1.groupBy('name').min().show()
```

```python
## Group by to find which departent gives max salary
df_pyspark1.groupBy('Department').sum().show()
```

```python
## Group by to find  departent mean salary
df_pyspark1.groupBy('Department').mean().show()
```

```python
## Group by to find  no of people in department
df_pyspark1.groupBy('Department').count().show()
```

```python
# Direct aggregate without using group by function

df_pyspark1.agg({'Salary':'sum'}).show()
```
