```
# File location and type
file_location = "/FileStore/tables/tips.csv"
file_type = "csv"

# The applied options are for CSV files. For other file types, these will be ignored.
df =spark.read.csv(file_location,header=True,inferSchema=True)
df.show()
```

```
df.printSchema()
```

```
## Here total_bill is dependent Variable
## Remaing all are independent variables , Here all these are categorical, so need to convert to numerical
```

```
df.columns
```

```
## Handling Categorical Feature

from pyspark.ml.feature import StringIndexer

# Here it creates new ordinal col
indexer = StringIndexer(inputCols = ['sex', 'smoker', 'day', 'time'], outputCols = ['sex_indexed', 'smoker_indexed', 'day_indexed', 'time_indexed'])

#Shift + tab to see parameter req for any function

df_r = indexer.fit(df).transform(df)
df_r.show()
```

```
df_r.columns
```

```
# Now Using VectorAssembler, we group all independent feature together and dependent feature separate
from pyspark.ml.feature import VectorAssembler
featureassembler = VectorAssembler(inputCols = ['tip', 'size', 'sex_indexed', 'smoker_indexed', 'day_indexed', 'time_indexed'], outputCol = 'Independent Feature')
output = featureassembler.transform(df_r)
output.show()
```

```
output.select('Independent Feature').show()
```

```
# Finalized data
finalized_data = output.select('Independent Feature','total_bill')
```

```
## Apply Traintest split and apply Linear Regression Model

from pyspark.ml.regression import LinearRegression

# Train test split
train_data, test_data = finalized_data.randomSplit([0.75,0.25])
test_data.show()
```

```
#In Pyspark, during model initialization time only input(featuresCol) and output(labelCol) col are defined. So no need to split like train_X,train_Y like pandas
regressor = LinearRegression(featuresCol = 'Independent Feature', labelCol = 'total_bill' )

#Fit the model
regressor = regressor.fit(train_data)
```

```
## coefficients

regressor.coefficients
#In regression, no of coefficients = no of input feature in same order

# Here for combined 1 col also it treats as separate col and provides its coefficient value
## 'tip', 'size', 'sex_indexed', 'smoker_indexed', 'day_indexed', 'time_indexed'
```

```
## Intercept, This is a coefficient value
regressor.intercept
```

```
# Evaluate our model

pred_results = regressor.evaluate(test_data)
```

```
## Final Comparison
pred_results.predictions.show()
# In Pyspark, it combines predicted o/p with orignal data and provide, but in Normal pandas it wont like this
```

```
## Evaluation

### R2, MAE, MSE, 
pred_results.r2, pred_results.meanAbsoluteError, pred_results.meanSquaredError
```
