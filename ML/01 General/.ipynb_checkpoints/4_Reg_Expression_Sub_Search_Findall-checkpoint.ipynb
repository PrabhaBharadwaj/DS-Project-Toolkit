## Regular expressions

## 1. SUB   --> SUBSTITUTE

```python
#re.sub(pattern, replacement_str, input_str): returns a new string
# it subtitutes the value at all the position
# Just string means whole character string it searches
import re
re.sub('abc', '*', 'abcdef abcdef')
```

```python
#List means any individual character in list are replaced
# replace a or b or c as a single letter
re.sub('[abc]', '*', 'adefbc cdefab')
```

```python
#2 list means Combination of each str in both list will search and replaces to *
# a1,a2,a3,a4,b1,b2,b3,b4,c1,c2,c3,c4
re.sub(r'[abc][1234]', '*', 'a1 + b3 + d4 + e5')
```

```python
# . dot means match any character except new line('\n')
re.sub(r'A.B', '*', 'A2B AxB AxxB, A$B')
```

```python
# ^ operator is used for check the starting of a string
re.sub(r'^abc' , '*', 'abcdeff abc abc')
```

```python
# $ operator is used for check the end of a string
re.sub(r'abc$' , '*', 'abcdeff abc abc')
```

```python
input_str = input("enter the string:")
re.sub(r'abc$' , '*', input_str)
```

### metacharacters
- characters having special meaning
- \d is used to match numbers
- \D matches non-digit
- \w matches alphanumeric character [a-zA-Z0-9_]
- '+' makes search greedy: 1 or more


```python
#Replaces only character
re.sub(r'\d', '*', 'my number is 8737837478879')
```

```python
#Replaces only non integer   [a-zA-Z]
re.sub(r'\D', '*', 'my number is 8737837478879')
```

```python
#Replaces any integer or character  [a-zA-Z0-9_]
re.sub(r'\w', '*', 'my number is 8737837478879')
```

## 2. SEARCH   --> Search

## re.search(pattern, input_str) 
- returns first matched value
- returns match object
- to see the match valued call .group() on the match object

```python
input_str = "my number is 8737837478879"
#It returns 1st matched object
match_obj = re.search(r'\d', input_str)
print(match_obj)
if match_obj:
    print(match_obj.group())
else:
    print("No match")
```

```python
# how to match the complete number:
# '+' makes search greedy: 1 or more till it reaches non numeric
input_str = "my number is 8737837478879 and 1234556"
match_obj = re.search(r'\d+', input_str)
print("matched obj", match_obj)
if match_obj:
    print(match_obj.group())  # to print the matched value
    print(match_obj.span()) # It gives match no positions in whole string
else:
    print("No match")
```

## 3. FINDALL   --> Findall

## re.findall(pattern, input_str)
- returns all the matches in a list of strings

```python
# how to match the complete number:
# '+' makes search greedy: 1 or more
input_str = "my number is 8737837478879 and 1234556"
match_values = re.findall(r'\d+', input_str)
print(match_values)
```

```python
# how to match the complete number:
# '+' makes search greedy: 1 or more
# d{3,5} means , match only the nos having atleast 3 digits and pick upto
# ...5 digits so any no less than 3 digits, would be ignored.
input_str = "my 12 and 45 is 8737837478879 and 1234556and 67845 and 78796969"
match_values = re.findall(r'\d{3,5}', input_str)
print(match_values)
```

# 4. Replace --> replace

```python
#6.4: Create a series and replace either a or dog with XX-XX'
    #A', 'B', 'C', 'AabX', 'BacX','', np.nan, 'CABA', 'dog', 'cat'
    
import re
import pandas as pd
import numpy as np
new_series3 = pd.Series(['A', 'B', 'C', 'AabX', 'BacX','', np.nan, 'CABA', 'dog', 'cat'])
print(new_series3.values)

# Here , case=False will replace both small a and big A
print(new_series3.str.replace('a|dog', 'XX-XX' , case=False) )
```

```python
# 6.5: Create a series and remove dollar from the numeric values
    # '12', '-$10', '$10,000'
    
new_series4 = pd.Series(['12', '-$10', '$10,000'])
print(new_series4.values)
print(new_series4.str.replace(r'$', '') )
```
