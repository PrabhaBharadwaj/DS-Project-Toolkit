# GroupBy operations

```python
dic = {"treatment":["A","A","B","B"],"gender":["F","M","F","M"],"response":[10,45,5,9],"age":[15,4,72,65]}
df = pd.DataFrame(dic)
df
```

```python
# according to treatment take means of other features
df.groupby("treatment").mean()   # mean is aggregation / reduction method
# there are other methods like sum, std,max or min
```

```python
# we can only choose one of the feature
df.groupby("treatment").age.max() 
```

```python
# Or we can choose multiple features
df.groupby("treatment")[["age","response"]].min() 
```

```python
print(df.info())
# as you can see gender is object
# However if we use groupby, we can convert it categorical data. 
# Because categorical data uses less memory, speed up operations like groupby
df["gender"] = df["gender"].astype("category")
df["treatment"] = df["treatment"].astype("category")
df.info()
```

## Aggregations:  Group by with  count
    An aggregated function returns a single aggregated value for each group. 
    Once the group by object is created, several aggregation operations can be performed on the grouped data.
    an obvious one is aggregation via the aggregate or equivalent agg method

```python
#To make group on Team, and show each groups row count as next col
#method 1
grouped = df.groupby('Team')
print(grouped.groups)

new_df = grouped.size().reset_index(name='counts')  #Here size will do count(*), reset_index will rename new count col to 'counts'
print(new_df)
print(type(new_df))

##OR

print("\n")
new_df = df['Team'].value_counts()  #Here value_counts() also provides count by grouping 'team without grp by function'
print(new_df)
print(type(new_df))
```

```python
#method 2 using agg function
grouped = df.groupby('Team')
print(grouped.groups)

#Here 1st group is done on team col then col 'rank' values mean are calculated for each group
print("\n Mean for col rank only")
new_df = grouped['Rank'].agg(np.mean)  #Here mean will do mean value by considering all rows in that grp
print(new_df)
print("#"*10)

#Here 1st group is done on team col then for all remaining numerical col's  mean are calculated for each group
print("\n Mean for all numerical col" )
new_df = grouped.agg(np.mean)  
print(new_df)
print("#"*10)

#Here 1st group is done on team col then count for each group. i.e row count
print("\n Count for all numerical col" )
new_df = grouped.agg(np.size)  #
print(new_df)
```

```python
grouped = df.groupby('Team')


#Here 1st group is done on team col then col 'rank' values mean are calculated for each group
print("\n sum, std, Mean for col rank only")
new_df = grouped['Rank'].agg([np.sum, np.std, np.mean])  #Here size will do count(*),
print(new_df)
```

```python
#Hierarchical Indexes:  We can groupby different levels of a hierarchical index using the level parameter:
#level 0 means group on 1st col
#level 1 means group on 2nd col

arrays = [['Falcon', 'Falcon', 'Parrot', 'Parrot'],
        ['Capitve', 'Wild', 'Capitve', 'Wild']]
# way to Define multiindesx by using function 'pd.MultiIndex.from_arrays'
index = pd.MultiIndex.from_arrays(arrays, names=('Animal', 'Type'))
print(index)
print("#"*10)
df1 = pd.DataFrame({'Max Speed' : [390., 350., 30., 20.]},   index=index)
print(df1)
print("#"*10)
newdf1 = df1.groupby(level = 0).mean()
print(newdf1)

print("#"*10)
newdf1 = df1.groupby(level = 1).mean()
print(newdf1)
```

```python
#Group based on one field and find mean of numeric field in that group

df1 = pd.DataFrame({'Animal' : ['Falcon', 'Falcon','Parrot', 'Parrot'],
                    'Max Speed' : [380., 370., 24., 26.],
                   'Oth Speed' : [10., 20., 30., 40.]})
print(df1)
df1.groupby(['Animal']).mean()
```

```python
#To get grouping value, we can use .index, .values for values
dn = df1.groupby(['Animal'])["Max Speed"].sum()
print(dn)
print("\n Print groups as index)")
print(dn.index)
print("\n Print other values after group)")
print(dn.values)
```

## Filter : Groupby then filteer 
    Filtration filters the data on a defined criteria and returns the subset of data. The filter() function is used to filter the data.

```python
import pandas as pd
import numpy as np

ipl_data = {'Team': ['Riders', 'Riders', 'Devils', 'Devils', 'Kings',
   'kings', 'Kings', 'Kings', 'Riders', 'Royals', 'Royals', 'Riders'],
   'Rank': [1, 2, 2, 3, 3,4 ,1 ,1,2 , 4,1,2],
   'Year': [2014,2015,2014,2015,2014,2015,2016,2017,2016,2014,2015,2017],
   'Points':[876,789,863,673,741,812,756,788,694,701,804,690]}
df = pd.DataFrame(ipl_data)
print("Full datframe \n" , df)

print("\n" , df.groupby('Team').groups)

# In the Below filter condition, we are asking to return the teams which have participated three or more times in IPL.
#Here groupby's each group will be passed to iteration to lambda func as x, 
#each groups len(x) means row count is considered to filter remaining groups/rows

new_df = df.groupby('Team').filter(lambda x: len(x) >= 3)
print("\n Filtered datframe \n" )
print(new_df)
```

## Concatenation  & Append
- process of combining two or more data structures
- Concatenation basically glues the dataframes together. You can select the dimension on which you want to concatenate.

- default axis = 0, i.e at the end of all coumns we did concatenation, if axis=1 concatenate will add 2nd DF as new columns

```python
import pandas as pd

world_champions={'Team':['India','Australia','West Indies','Pakistan','Sri Lanka'],
           'ICC_rank':[2,3,7,8,4],
           'World_champions_Year':[2011,2015,1979,1992,1996],
           'Points':[874,787,753,673,855]}

chokers={'Team':['South Africa','New Zealand','Zimbabwe'],
           'ICC_rank':[1,5,9],
           'Points':[895,764,656]}
df1 = pd.DataFrame(world_champions)
df2 = pd.DataFrame(chokers)
print("DF1: World champions")
print(df1)
print("#"*20)
print("DF2: Chokers")
print(df2)
```

```python
# concat()
concatenated_df = pd.concat([df1, df2], sort=True)
concatenated_df
```

```python
# concat()
#Default axis = 0, i.e at the end of all coumns we did concatenation
concatenated_df1 = pd.concat([df1, df2], sort=True, axis = 0)
concatenated_df1
```

```python
# concat()
# axis = 1 at the end of all rows we did concatenation
concatenated_df2 = pd.concat([df1, df2], sort=True, axis = 1)
concatenated_df2
```

### Append:

- same as concat
- default axis = 0, i.e at the end of all coumns we did concatenation, if axis=1 concatenate will add 2nd DF as new columns

```python
# 2.concantenation can also be done with the df.append()

import pandas as pd

world_champions={'Team':['India','Australia','West Indies','Pakistan','Sri Lanka'],
           'ICC_rank':[2,3,7,8,4],
           'World_champions_Year':[2011,2015,1979,1992,1996],
           'Points':[874,787,753,673,855]}

chokers={'Team':['South Africa','New Zealand','Zimbabwe'],
           'ICC_rank':[1,5,9],
           'Points':[895,764,656]}
df1 = pd.DataFrame(world_champions)
df2 = pd.DataFrame(chokers)
print("DF1: World champions")
print(df1)
print("#"*20)
print("DF2: Chokers")
print(df2)

print("********** Appending/concatenating *******")
new_df = df1.append(df2, sort=True)
print(new_df)
```

## Merging and Joining

   ### Merging: 
   In merging, you can merge two data frames to form a single data frame. You can also decide which columns you want to make common 
   ### Join: 
   This is quite similar to the “merge” operation, except the joining operation will be on the “index” instead of  the “columns”.

```python
##Merge

# inner join returns intersection of two Dataframes.
#Merge is like a join in sql, it merges 2 dataset and list all columns next to each other
world_champions={'Team':['India','Australia','West Indies','Pakistan','Sri Lanka'],
           'ICC_rank':[2,3,7,8,4],
           'World_champions_Year':[2011,2015,1979,1992,1996],
           'Points':[874,787,753,673,855]}

chokers={'Team':['India','New Zealand','Zimbabwe'],
           'ICC_rank':[2,5,9],
           'Points':[874,764,656]}
df1 = pd.DataFrame(world_champions)
df2 = pd.DataFrame(chokers)
print("DF1: World champions")
print(df1)
print("#"*20)
print("DF2: Chokers")
print(df2)
print(" *******Default is Inner Join Merge*********")
print(pd.merge(df1, df2, on='Team'))
```

```python
# right join merges two dataframes based on the keys of the right dataframe
world_champions={'Team':['India','Australia','West Indies','Pakistan','Sri Lanka'],
           'ICC_rank':[2,3,7,8,4],
           'World_champions_Year':[2011,2015,1979,1992,1996],
           'Points':[874,787,753,673,855]}

chokers={'Team':['India','New Zealand','Zimbabwe'],
           'ICC_rank':[2,5,9],
           'Points':[874,764,656]}
df1 = pd.DataFrame(world_champions)
df2 = pd.DataFrame(chokers)
print("DF1: World champions")
print(df1)
print("#"*20)
print("DF2: Chokers")
print(df2)
print(" *******Right Join Merge*********")
print(pd.merge(df1, df2, on='Team', how='right'))
```

```python
# left join merges two dataframes based on the keys of the left dataframe
world_champions={'Team':['India','Australia','West Indies','Pakistan','Sri Lanka'],
           'ICC_rank':[2,3,7,8,4],
           'World_champions_Year':[2011,2015,1979,1992,1996],
           'Points':[874,787,753,673,855]}

chokers={'Team':['India','New Zealand','Zimbabwe'],
           'ICC_rank':[2,5,9],
           'Points':[874,764,656]}
df1 = pd.DataFrame(world_champions)
df2 = pd.DataFrame(chokers)
print("DF1: World champions")
print(df1)
print("#"*20)
print("DF2: Chokers")
print(df2)
print(" *******Left Join Merge*********")
print(pd.merge(df1, df2, on='Team', how='left'))
```

```python
# outer join merges two dataframes based on the full union
world_champions={'Team':['India','Australia','West Indies','Pakistan','Sri Lanka'],
           'ICC_rank':[2,3,7,8,4],
           'World_champions_Year':[2011,2015,1979,1992,1996],
           'Points':[874,787,753,673,855]}

chokers={'Team':['India','New Zealand','Zimbabwe'],
           'ICC_rank':[2,5,9],
           'Points':[874,764,656]}
df1 = pd.DataFrame(world_champions)
df2 = pd.DataFrame(chokers)
print("DF1: World champions")
print(df1)
print("#"*20)
print("DF2: Chokers")
print(df2)
print(" *******outer Join Merge*********")
print(pd.merge(df1, df2, on='Team', how='outer'))
```

```python
# inner join returns intersection of two Dataframes.
world_champions={'Team':['India','Australia','West Indies','Pakistan','Sri Lanka'],
           'ICC_rank':[2,3,7,8,4],
           'World_champions_Year':[2011,2015,1979,1992,1996],
           'Points':[874,787,753,673,855]}

chokers={'Team':['India','New Zealand','Zimbabwe'],
           'ICC_rank':[2,5,9],
           'Points':[874,764,656]}
df1 = pd.DataFrame(world_champions)
df2 = pd.DataFrame(chokers)
print("DF1: World champions")
print(df1)
print("#"*20)
print("DF2: Chokers")
print(df2)
print(" *******Inner Join Merge*********")
print(pd.merge(df1, df2, on='Team', how='inner'))
```

```python
# 6.10: Create pandas dataframe having keys and ltable and rtable as below -
    # 'key': ['One', 'Two'], 'ltable': [1, 2]
    #  'key': ['One', 'Two'], 'rtable': [4, 5]
    # Merge both the tables based of key

ltable = {'key': ['One', 'Two'], 'ltable': [1, 2]}
rtable = {'key': ['One', 'Two'], 'rtable': [4, 5]}

new_pd1 = pd.DataFrame(data=ltable)
new_pd2 = pd.DataFrame(data=rtable)
print("Newpd1: \n " , new_pd1)
print("Newpd2: \n " , new_pd2)

#Method 1
new_pd3 = pd.merge(new_pd1, new_pd2, on = 'key')
print("Newpd3: \n " , new_pd3)
```

   ### Join: 
   This is quite similar to the “merge” operation, except the joining operation will be on the “index” instead of  the “columns”.

```python
df1 = pd.DataFrame({"Int_Rate":[2,1,2,3], "IND_GDP":[50,45,45,67]}, index=[2001, 2002,2003,2004])

print(df1)
 
df2 = pd.DataFrame({"Low_Tier_HPI":[50,45,67,34],"Unemployment":[1,3,5,6]}, index=[2001, 2003,2004,2004])

print(df2)
 
joined= df1.join(df2)
print(joined)
```

# CATEGORIAL

     this will save spaces, like TD compress.
     We have datafrma with sex as Male1 ,male2,male3, fmale1, fmale2 ,fmale3 , by using datatype=categorial, it convers it into category with some integer value

```python
#cATEGORIAL , this will save spaces, like TD compress
# We have datafrma with sex as Male1 ,male2,male3, fmale1, fmale2 ,fmale3 , by using datatype=categorial, it convers it into category with some integer value
df = pd.DataFrame({"sex": ['male1' ,'male2','male1', 'fmale1', 'fmale2' ,'fmale1'], "Name": ['wastu','Pranamya','Pallu', 'Pavani','me','kanda']})
df
```

```python
df.sex = pd.Categorical(df.sex)
df
#Just with above it wont assign
```

```python
#here cat.codes will convert similar value to 1 category
df.sex.cat.codes
```

```python
df.sex = df.sex.cat.codes
df
# male1  -->2 ,male2  -->3 ,fmale1  -->0 ,fmale2  -->1
```
