## List comprehension
- Def: creating a new list from an existing one.


```python
# Syntax[<operation> for i in <old_list> <filter_condition>]
# <filter_condition> is optional
a = [1, 2, 3, 4]
# target O/p: [1, 4, 9, 16]

b = [i**2 for i in a]
b
```

```python
a = [1, 2, 3, 4, 5, 6, 7]
# target O/p: [1, 4, 9, 16]
b = [i**2 for i in a if i<5]
print(b)

### same thing in longer code
a = [1, 2, 3, 4, 5, 6, 7]
b = []
for i in a:
    if i<5:
      b.append(i**2)  
print(b)
```

```python
#11. By using list comprehension, please write a program to print the 
#list after removing delete numbers which are divisible by 5 and 7 in [12,24,35,70,88,120,155]
a = [12,24,35,70,88,120,155]

b = [i for i in a if i%5 != 0 and i%7 != 0]
print(b)
```
