```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# load the data
data = pd.read_csv("movie_metadata1.csv")
data.head()
```

```python
# getting the relevant values from the input dataset
f1 = data['director_facebook_likes'].values
f2 = data['actor_3_facebook_likes'].values
#X = data[['director_facebook_likes', 'actor_3_facebook_likes']].values
X = np.array(list(zip(f1,f2)))
plt.scatter(f1,f2, c='black', s=50)
```

```python
# Eucledian distance calculator
def dist(a,b,ax=1):
    return np.linalg.norm(a-b,axis=ax)
k = 5

# 1: random centroids
# X coordinate of random centroids
C_x = np.random.randint(0, np.max(X)-20, size=k)
# y coordinate of random centroids
C_y = np.random.randint(0, np.max(X)-20, size=k)

# random centroids
C = np.array(list(zip(C_x, C_y)), dtype=np.float32)
C
```

```python
# plotting data points along with the centroid
plt.scatter(f1,f2,s=10)
plt.scatter(C_x, C_y, marker='*', s=200, c='g')
```
