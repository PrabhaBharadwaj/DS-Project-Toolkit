# PACIFIC HURRICANE

> The NHC publishes the tropical cyclone historical database in a format known as HURDAT, short for HURricane DATabase. These databases (Atlantic HURDAT2 and NE/NC Pacific HURDAT2) contain six-hourly information on the location, maximum winds, central pressure, and (starting in 2004) size of all known tropical cyclones and subtropical cyclones.


> Here STATUS is output column, based on other feature we need to decide which is the best model to predict Hurricane status

> Its categorial o/p not continous one. So its not a regression problem. Its a classification problem with more than 2 value to predict. So we use Supervised Classification Models like 

> Decision Tree

> Random Forest

##  1. Import all Librariy

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn import tree
from sklearn import metrics

from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report

from sklearn.metrics import roc_curve
from sklearn.metrics import auc
```

```python
# loading the data
data = pd.read_csv('../Dataset/pacific.csv')
data.head()
```

## 2. Data Exploration with Graphs and Visualization

```python
#removing Hemisphere tags from coordiantes:
#It right side trims W,E from Longitude column and N , S from Latitude column
data['Longitude'] = data['Longitude'].map(lambda x: x.rstrip('W'))
data['Latitude'] = data['Latitude'].map(lambda x: x.rstrip('N'))
data['Latitude'] = data['Latitude'].map(lambda x: x.rstrip('S'))
data['Longitude'] = data['Longitude'].map(lambda x: x.rstrip('E'))

pred_columns.head()
```

```python
#converting Latitude and Longitude to floats:
data['Latitude'] = data['Latitude'].astype(float)
data['Longitude'] = data['Longitude'].astype(float)
data.info()
```

```python
# data  wrangling, Remove unnecessary column
pred_columns = data[:]
pred_columns.drop(['ID','Name','Event','Status'],  #'Latitude','Longitude'
                 inplace=True,
                 axis=1)
pred_columns.head()
```

# 3. Building Machine Learning Decision Tree Models

```python
# split the data
train, test = train_test_split(data, test_size=0.3)
print(train.shape)
print(test.shape)
```

```python
train_X = train[relevant_columns]
train_Y = train['Status']

test_X = test[relevant_columns]
test_Y = test['Status']
```

## APPLY Label encoder to convert op category to integer

> In Decision tree and random  forest will supports categorial output. It wont have much difference with label encoding. But still used here

```python
from sklearn import preprocessing
le = preprocessing.LabelEncoder()
#Or we can use fit_transform instead of separate fit and transform
#le.fit_transform(Y)

# Fit Applied on whole set, few category may miss in train data, so used full
le.fit(train['Status'])
#le.classes_ provides distinct labels
print(le.classes_)
##### 0 -->B,  1--->M

#This converts data from  B -->0 aand M to 1   means here its normalized  in 0,1
train_Y = le.transform(train_Y)
train_Y
test_Y = le.transform(test_Y)
#test_Y
```

```python
# train the model
dt = tree.DecisionTreeClassifier()
dt.fit(train_X, train_Y)

#testing the model
prediction_dt_op = dt.predict(test_X)

# Do inverse
prediction_dt_op1 = le.inverse_transform(prediction_op)
test_Y1 = le.inverse_transform(test_Y)

# measure the accuracy
metrics.accuracy_score(test_Y, prediction_dt_op)
```

# 4. Building Machine Learning Random Forest Models

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(n_estimators=100)
# train the model
rf.fit(train_X, train_Y)


#testing
prediction_rf_op = rf.predict(test_X)

# Do inverse
prediction_rf_op1 = le.inverse_transform(prediction_rf_op)
test_Y1 = le.inverse_transform(test_Y)

metrics.accuracy_score(test_Y, prediction_rf_op)
```

```python
from sklearn.multiclass import OneVsRestClassifier
from sklearn.ensemble import RandomForestClassifier


n_classes = 12 # It should be no of distinct classes in output

# classifier
clf = OneVsRestClassifier(RandomForestClassifier(n_estimators=100))
#y_score = clf.fit(train_X, train_Y).decision_function(test_X)
clf.fit(train_X, train_Y)
pred = clf.predict(test_X)
pred_prob = clf.predict_proba(test_X)


# Compute ROC curve and ROC area for each class
fpr = dict()
tpr = dict()
roc_auc = dict()
for i in range(n_classes):
    fpr[i], tpr[i], _ = roc_curve(test_Y1[:, i], pred_prob[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])
```

```python

# Plot of a ROC curve for a specific class
for i in range(n_classes):
    plt.figure()
    plt.plot(fpr[i], tpr[i], label='ROC curve (area = %0.2f)' % roc_auc[i])
    plt.plot([0, 1], [0, 1], 'k--')
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('Receiver operating characteristic example')
    plt.legend(loc="lower right")
    plt.show()
```
