#### 2.2.1 CORRELATION - Here Write a function to take only selected column which is having correlation above threshold

```python

def getCorrelatedFeature(corrdata, threshold):
    feature = []  # Col name
    value = []  ## Correlation value
    
    for i, index in enumerate(corrdata.index):  ## here i =0,1,2,3  index ='CRIM', 'ZN', 'INDUS', 'CHAS', 'NOX',etc
        if abs(corrdata[index])> threshold:   # pls note we have taken here absolute value.    0.388305>0.50 
            feature.append(index)
            value.append(corrdata[index])
            
    df = pd.DataFrame(data = value, index = feature, columns=['Corr Value'])
    return df
```

```python
# It returns column which has correlation above 0.50 with correlation value 'shares'
threshold = 0.050 #5%
corr_value = getCorrelatedFeature(correlation['shares'], threshold)
corr_value
```

- By seeing data we come to know none of the field has good correlation with output field shares.

- So we cant choose best field having correlation above 50%, all are less than 5% correlation only.

### 2.3 SCALARIZE DATA:  

#### 2.3.1 STANDARD SCALARA MEDHOD: All are float data, so do scalarize the data
- It converts all field with single unit varience (0 mean 1 varience)

```python
from sklearn.preprocessing import StandardScaler

## Here data not splitted, directly applied SCalar and PCA

scaler = StandardScaler()

# Fit on training set only.
scaler.fit(Social_data_X)
# scaler.fit(train_X)

# Apply transform to both the training i/p set and the test i/p set. or whole input set
#transform 
Social_data_X1 = scaler.transform(Social_data_X)
# scaler.transform(train_X)
# scaler.transform(test_X)

Social_data_X1
```

```python
# Implementing the PCA and resducing the number of columns
from sklearn.decomposition import PCA

for i in range(1,6):
    pca_model = PCA(n_components=i)
    pca_model.fit(Social_data_X1)

    # transform the data from multidimension to 2D
    data_2d = pca_model.transform(Social_data_X1)
    data_df_2d = pd.DataFrame(data_2d)
    a = pca_model.explained_variance_ratio_
    print("n component = %s total varience is %s " %( i, a.sum()))
```

```python
# Model complexity
neig = np.arange(1, 50)
train_accuracy = []
test_accuracy = []
# Loop over different values of k
for i, k in enumerate(neig):
    # k from 1 to 25(exclude)
    knn = KNeighborsClassifier(n_neighbors=k)
    # Fit with knn
    knn.fit(train_X,train_Y)
    #train accuracy
    train_accuracy.append(knn.score(train_X, train_Y))
    # test accuracy
    test_accuracy.append(knn.score(test_X, test_Y))

# Plot
plt.figure(figsize=[13,8])
plt.plot(neig, test_accuracy, label = 'Testing Accuracy')
plt.plot(neig, train_accuracy, label = 'Training Accuracy')
plt.legend()
plt.title('-value VS Accuracy')
plt.xlabel('Number of Neighbors')
plt.ylabel('Accuracy')
plt.xticks(neig)
plt.savefig('graph.png')
plt.show()

# Here it gives best K value with max accuracy for test data
print("Best accuracy is {} with K = {}"
      .format(np.max(test_accuracy),1+test_accuracy.index(np.max(test_accuracy))))
```

- n_neighbors=46 having less distortion. this kept as k

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=46)
knn.fit(train_X, train_Y)
KNN_pred_y = knn.predict(test_X)
# compare the house price predicted vs known test value
data = {'predicted O/P':KNN_pred_y,'known(testing) O/P':test_Y}
pd.DataFrame(data)
print("Score: " ,knn.score(test_X,test_Y))

print("KNN Accuracy: " ,metrics.accuracy_score(test_Y, KNN_pred_y))
```
