- Another kind of regularized regression that you ,could use instead of ridge regression is called Lasso Regression.

- Lasso regression uses a slightly different regularisation term called an L1 penalty, instead of ridge regression's L2 penalty .
- Lasso regression is most helpful if you think there are only a few variables that have a medium or large effect on the output variable. 

-The L1 penalty looks kind of similar to the L2 penalty, in l2 it computes a sum of square of the coefficients but it's some of the absolute values of the W-coefficients in L1
