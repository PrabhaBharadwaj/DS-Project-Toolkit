- Normally Linear regression provides straigt line based on training dataset, but some time this line will overfits training data and 0 variance and large varience for testing datsets.

- In ridge regression it produces new line, it wont overfit training data, it will have litte bias in training data and little varience for test data compared old line which had huge varience

- In  Ridge regression, it will have least square(linear regression) equation, with extra 2 parameter which is called as lambda i.e  penalty

- If there are lots of variables that contribute small or medium effects on output variable, there ridge regression is typically the better choice.

- Lasso regression is most helpful if you think there are only a few variables that have a medium or large effect on the output variable. 

- The amount of regularisation to apply is controlled by the alpha parameter.Larger alpha means more regularization and simpler linear models with weights closer to zero. 

-The L2 penalty looks kind of similar to the L1 penalty, in L1 it computes  absolute values of the W-coefficients but it's sum  of the squares of the W-coefficients in L2.



- Ridge regression uses the same least-squares criterion, but with one difference. During the training phase, it adds a penalty for feature weights,

- This addition of a penalty term to a learning algorithm's objective function is called Regularisation. It's a way to prevent overfitting, and thus, improve the likely generalization performance of a model, by restricting the models possible parameter settings.  

- Regression problems with dozens or hundreds of features, the accuracy improvement from using regularized linear regression like ridge regression could be significant. 




- The amount of regularisation to apply is controlled by the alpha parameter.

- Larger alpha means more regularization and simpler linear models with weights closer to zero. 
The default setting for alpha is 1.0. 


###### Alpha:


- Ridge regression is regularizing the linear regression by imposing that sum of squares penalty on the size of the W coefficients.

- So the effect of increasing alpha is to shrink the AW coefficients toward zero and towards each other.
But if the input variables, the features, have very different scales, then when this shrinkage happens of the coefficients, input variables with different scales will have different contributions to this L2 penalty, because the L2 penalty is a sum of squares of all the coefficients. 

- If scales are different, So transforming the input features, so they're all on the same scale by using feature normalization.

- MinMax Scaling is one of the scalar normalization method. that will transform all the input variables, so they're all on the same scale between zero and one

- In general, regularisation works especially well when you have relatively small amounts of training data compared to the number of features in your model.
