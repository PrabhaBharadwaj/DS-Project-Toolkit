# PCA

## PCA
> PCA is a technique which helps us in extracting a new set of variables from an existing large set of variables. These newly extracted variables are called Principal Components.

> In PCA, algorithm we are not doing any prediction. Here we are reducing the driver/input columns in dataset


> If it has more no of highly correlated or similar kind data, then apply PCA/LDA on I/P data set to reduce the dimension of the dataset(Dimension Reduction Tech)

> Before applying PCA/LDA on I/P data, convert data to more scalar way(Scale data so that each feature has a single unit variance.) by using "StandardScaler" approach

> PCA applying time all col will be normally similar and numeric, some time index will have value col, ex: country as index and year wise sale data as columns. If not we can convert like that.

> By using PCA, we will find PC1, PC2 etc which are new datasets, these are the  underlaying causes/Drivers

#### Eigen Vector

> By using Eigen Decomposition, Eigen vector we will decide no of PCA component


## 1. Import Library

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("../Dataset/trans_us_original.csv",
                   index_col=0,  # Read 1st col as index
                  thousands=',')
#data.iloc[96] # value having comma, so used thousands=','
```

```python
data.head()
```

```python
data.index.names = ['countries']
data.columns.names = ['years']
data.head()
```

## 2. APPLY STANDARD SCALAR

- Before apply PCA function better to apply  scaler on data

- Scale data so that each feature has a single unit variance. Without splitting data as test and train

- Generally scale will be applied on Input dataset

###  Standardize the Data
   -  PCA is effected by scale so you need to scale the features in the data before applying PCA. 
    - You can transform the data onto unit scale (mean = 0 and variance = 1) which is a requirement for the optimal performance of many machine learning algorithms.

    
    
### StandardScaler

 - This helps standardize the dataset’s features. Note you fit on the training set and transform on the training and test set.
       Refer Modul8 case study 3, here its used 


```python
from sklearn.preprocessing import StandardScaler

## Here data not splitted, directly applied SCalar and PCA

scaler = StandardScaler()

# Fit on training set only.
scaler.fit(data)
# scaler.fit(train_X)

# Apply transform to both the training i/p set and the test i/p set. or whole input set
#transform 
scaled_data = scaler.transform(data)
# scaler.transform(train_X)
# scaler.transform(test_X)

scaled_data
```

## 3. APPLY PCA

### 3.1 principl component 2

> Here we are specifing no of component implicitly. we can give as percentage also

```python
# Implementing the PCA and resducing the number of columns
from sklearn.decomposition import PCA

pca_model = PCA(n_components=2)  # Here we give no of components
#or 
#pca_model = PCA(.95)  # Here we give varience in %, so component will be decided automatically

#train the model
pca_model.fit(scaled_data)

# transform the data from multidimension to 2D
data_2d = pca_model.transform(scaled_data)
data_df_2d = pd.DataFrame(data_2d)
data_df_2d.head()
```

```python
# do some formatting, APPEND INDEX BAME OF ORIGINAL DATA TO NEW DATA AND RENAME COLNAME
data_df_2d.index = data.index
data_df_2d.columns = ['PC1', 'PC2']
data_df_2d
```

```python
# variance
pca_model.explained_variance_ratio_
```

```python
## Meaning 1st one has 91.8% variance of original data so on
# PC1: 92.0 %
# PC2: 5.7 %
# = 97.5%
```

### 3.2 principl component 3

```python
# 3 principal components
# Implementing the PCA and resducing the number of columns
from sklearn.decomposition import PCA

pca_model = PCA(n_components=3)

#train the model
pca_model.fit(scaled_data)

# transform the data from multidimension to 2D
data_2d = pca_model.transform(scaled_data)
data_df_2d = pd.DataFrame(data_2d)

data_df_2d.head()
# do some formatting
data_df_2d.index = data.index
data_df_2d.columns = ['PC1', 'PC2', 'PC3']
data_df_2d.head()
```

## 4. Explained Variance
- The explained variance tells you how much information (variance) can be attributed to each of the principal components. This is important as  you can convert n dimensional space to 2 dimensional space, you lose some of the variance (information).

```python
# variance
pca_model.explained_variance_ratio_
```

```python
# PC1: 92.8 %
# PC2: 5.7 %
# PC3: 1 %
```
