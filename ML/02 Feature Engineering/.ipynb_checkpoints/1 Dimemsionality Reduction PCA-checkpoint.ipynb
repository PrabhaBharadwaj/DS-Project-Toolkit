# PCA

### 
> If it has more no of highly correlated or similar kind data, then apply PCA/LDA on I/P data set to reduce the dimension of the dataset(Dimension Reduction Tech)

```python
import pandas as pd
import matplotlib.pyplot as plt

```

```python
data = pd.read_csv("trans_us_original.csv",
                   index_col=0,
                  thousands=',')
#data.iloc[96] # value having comma
```

```python
data.head()
```

```python
data.index.names = ['countries']
data.columns.names = ['years']
data.head()
```

```python
# Implementing the PCA and resducing the number of columns
from sklearn.decomposition import PCA

pca_model = PCA(n_components=2)  # Here we give no of components
#or 
#pca_model = PCA(.95)  # Here we give varience in %, so component will be decided automatically

#train the model
pca_model.fit(data)

# transform the data from multidimension to 2D
data_2d = pca_model.transform(data)
data_df_2d = pd.DataFrame(data_2d)
data_df_2d.head()
```

```python
# do some formatting
data_df_2d.index = data.index
data_df_2d.columns = ['PC1', 'PC2']
data_df_2d
```

```python
# variance
pca_model.explained_variance_ratio_
```

```python
# PC1: 91.8 %
# PC2: 6 %
# = 98%
```

```python
# 3 principal components
# Implementing the PCA and resducing the number of columns
from sklearn.decomposition import PCA

pca_model = PCA(n_components=3)

#train the model
pca_model.fit(data)

# transform the data from multidimension to 2D
data_2d = pca_model.transform(data)
data_df_2d = pd.DataFrame(data_2d)
data_df_2d.head()
# do some formatting
data_df_2d.index = data.index
data_df_2d.columns = ['PC1', 'PC2', 'PC3']
data_df_2d.head()
```

```python
# variance
pca_model.explained_variance_ratio_
```

```python
# PC1: 91.8 %
# PC2: 6 %
# PC3: 1 %
```


###  Standardize the Data
    PCA is effected by scale so you need to scale the features in the data before applying PCA. 
    You can transform the data onto unit scale (mean = 0 and variance = 1) which is a requirement for the optimal performance of many machine learning algorithms.
   ### StandardScaler
       This helps standardize the dataset’s features. Note you fit on the training set and transform on the training and test set.
       Refer Modul8 case study 3, here its used 
       
       ### Before apply PCA
    scaler = StandardScaler()
    # Fit on training set only.
    scaler.fit(train_x)
    # Apply transform to both the training set and the test set.
    train_x = scaler.transform(train_x)
    test_x = scaler.transform(test_x)
