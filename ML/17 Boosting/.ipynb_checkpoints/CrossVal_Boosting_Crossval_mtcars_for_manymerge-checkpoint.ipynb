## Module 14 Cross Validation Model selection

### Boosting:	

- Concept of converting a group of week learner to strong learners is called Boosting.
- The term ‘Boosting’ refers to a family of algorithms which converts weak learner to strong learners.


- AdaBoost (Adaptive Boosting) :		
		
>  It fits a sequence of weak learners on different weighted training data.	

> It starts by predicting original data set and gives equal weight to each observation.

> If prediction is incorrect using the first learner, then it gives higher weight to observation which have been predicted incorrectly. 	

> Being an iterative process, it continues to add learner(s) until a limit is reached in the number of models or accuracy.	
    
> We can use AdaBoost algorithms for both classification and regression problem.

##  1. Import all Librariy

### CAR company data is there, based on its existing data, predict its feedback. which is the o/p.

- Data looks continous and o/p info is there, so we will go with supervised learning

##  1. Import all Librariy

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.neighbors import KNeighborsClassifier

from sklearn import metrics
```

```python
#Read data
df = pd.read_csv (r'..\Dataset\\mtcars_for_manymerge.csv')
x = df[['cyl','hp']]
y = df['Feedback']
df.head()
```

##  2. Apply KNN Classifier algorithm- Split data and apply  KNN with Random_state= 5

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,random_state =5)

knn = KNeighborsClassifier(n_neighbors=4)
knn.fit(x_train,y_train)
y_pred = knn.predict(x_test)

metrics.accuracy_score(y_test,y_pred)
```

##  2. Apply KNN Classifier algorithm- Split data and apply  KNN with Random_state= 10

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,random_state = 10)
knn = KNeighborsClassifier(n_neighbors=4)
knn.fit(x_train,y_train)
y_pred = knn.predict(x_test)
metrics.accuracy_score(y_test,y_pred)
```

## 3. Conclusion: 

> By random_state=10 gave lesser accuracy than random_satte=5, so just with data combination accuracy is varying. so we cant decide directly based on accuracy

## 4. To conclude this type of situtaion we will use Cross validation library.

- for this we will  set k-fold, it applys accuracy iteration on k times with different data combination and provides avg accuracy as o/p.
- This accuracy we can relay.
- This we will apply on different models and Based on this we will decide which is the best model


### 4.1 Cross validation on KNN model

```python
from sklearn.model_selection import cross_val_score

#KNN model assign
knn = KNeighborsClassifier(n_neighbors=4)

## Apply Cross validation on knn model, with k-fold/cv=10, and take mean which is the avg of all 10 accuracy
print(cross_val_score(knn, x, y, cv=10, scoring ='accuracy').mean())
```

### 4.2 Cross validation on Logistic regression model

```python
from sklearn.linear_model import LogisticRegression

#Logistic model assign

logreg = LogisticRegression()

## Apply Cross validation on logreg model, with k-fold/cv=10, and take mean which is the avg of all 10 accuracy
print (cross_val_score(logreg, x, y, cv=10, scoring = 'accuracy').mean())
```

## 5. Conclusion: 

> By cross validation, we can see KNN model gives better accuracy than logistic regression. so we can go with KNN for this problem model build
