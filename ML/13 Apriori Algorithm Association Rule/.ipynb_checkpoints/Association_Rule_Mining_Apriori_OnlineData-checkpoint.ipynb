> By applying Apriori algorithm and Association rules, we will get product combination like. If person buys prd1 then he had this much confidence to but prod2



##  1. Import all Librariy

```python
import pandas as pd

from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
```

```python
df = pd.read_excel(r'..\Dataset\Online_Retail.xlsx')
df.head()
```

> By seeing data invoice no are repeated in dataset with different product detal and count etc

## 2. Data Exploration

```python
df.info()
```

#### During this data wrangling, Invoice no having multiple product with its quantity, so invoice no made as index, products names as 1 column splitted a multiple columns with So that its easy to apply apriori algorithm.

> Reason: Many duplicates invoice and product, so its difficult to recommend


```python
# data wrangling

## This removes white spaces in both the sides, 
df['Description'] = df['Description'].str.strip()    

# dropna is used to delete row completely if col/subset ='InvoiceNo' is null/nan
# Here if invoice is nul, then deleteing whole row from dataset
df.dropna(axis=0, subset=['InvoiceNo'], inplace=True)
```

```python
# Converting datatype to str before apply 'str' contains funct
df['InvoiceNo'] = df['InvoiceNo'].astype('str')
```

```python
# Some invice contains value as 'C' that we are copied to df1
df1 = df[df['InvoiceNo'].str.contains('C')]
df1
```

```python

# Some invice contains value as 'C' that we need to remove completely
df = df[~df['InvoiceNo'].str.contains('C')]
df
```

> Before DF count:541909 rows × 8 columns.  Now DF count 532621 rows × 8 columns

```python
# consolidating different transactions based on country and description
#Convert 1 column (Description) values to no of columns based on its unique value and
# another col will be its value (Quantity)

#Here set_index('InvoiceNo') to index value
# Here Unstack will converts unique \"Description" value to different column
# fillna(0) --> for wach row value if ['Quantity'] are there will put that else puts value to 0 




basket = (df[df['Country'] == "France"].groupby(['InvoiceNo', 'Description'])['Quantity']
          .sum().unstack().reset_index().fillna(0)
          .set_index('InvoiceNo'))
basket
```

```python
# Here we have built one functiom encode_units, it convets count > 1 to default 1, if its lesser than 0 to 0
# Here we are ignoring each product count for that invoice, defaulting to 1 

def encode_units(x):
    if x <= 0:
        return 0
    if x >= 1:
        return 1
#Here apply map will apply condition on each value in df
basket_sets = basket.applymap(encode_units)

# Remove col 'POSTAGE'
basket_sets.drop('POSTAGE', inplace=True, axis=1)
basket_sets
```

## 3. Building Machine Learning Associaton Rule

### 3.1 Apply apriori Algorithm to get frequent_itemsets 

```python
# algorithm comes into picture
# generate frequent itemsets having minimum support value of 10%
frequent_itemsets = apriori(basket_sets, min_support=0.10, use_colnames=True)
frequent_itemsets
```

> By this we got 26 frequent_itemsets with min_support> 10% (>0.10)

> Here ALARM CLOCK BAKELIKE PINK  --> this is one frequent itemset. Its one whole product name

```python
basket_sets["ALARM CLOCK BAKELIKE PINK"].head()
```

### 3.2 Apply association_rules to get rules on frequent_itemsets .

```python
# Applied association_rules with min_threshold=1
rules = association_rules(frequent_itemsets, metric='lift', min_threshold=1)
rules
```

> By this we got combination of 26 frequent_itemsets as antecedents	consequents

> Here antecedents = (PLASTERS IN TIN CIRCUS PARADE), consequents = (PLASTERS IN TIN WOODLAND ANIMALS)

> Means if persons buys (PLASTERS IN TIN CIRCUS PARADE) then 0.606061 confidence that he buys (PLASTERS IN TIN WOODLAND ANIMALS)

## 4. Conclusion

### After applying Apriori algorithm and Association rule  we will get final set. On that based on our requirement we set confidence level and use that as the final set to target the user OR Keep those items nearby in store


```python
rules[(rules['lift'] >= 6) & (rules['confidence'] >= 0.8) ]
```

> Here antecedents = ((SET/6 RED SPOTTY PAPER PLATES)), consequents = (SET/20 RED RETROSPOT PAPER NAPKINS)

> Means if persons buys (SET/6 RED SPOTTY PAPER PLATES) then 0.8000 confidence that he buys (SET/20 RED RETROSPOT PAPER NAPKINS)
