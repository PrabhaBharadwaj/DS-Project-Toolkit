```python
# CANCER DETECTION
> We have cancer data , we need to prdict its Malignant or Benign
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn import tree
from sklearn import metrics
```

```python
# loading the data
data = pd.read_csv('./Datasets/pacific.csv')
data.head()
```

```python
sns.countplot(data['Status'])
plt.show()
```

```python
# data  wrangling
pred_columns = data[:]
pred_columns.drop(['ID','Name','Event','Status','Latitude','Longitude'],
                 inplace=True,
                 axis=1)
pred_columns.head()
```

```python
# store the relevant columns
relevant_columns = pred_columns.columns
print(list(relevant_columns))
```

```python
# split the data
train, test = train_test_split(data, test_size=0.3)
print(train.shape)
print(test.shape)
```

```python
train_X = train[relevant_columns]
train_Y = train['Status']

test_X = test[relevant_columns]
test_Y = test['Status']
```

```python
# train the model
model = tree.DecisionTreeClassifier()
model.fit(train_X, train_Y)

#testing the model
prediction_op = model.predict(test_X)

# measure the accuracy
metrics.accuracy_score(test_Y, prediction_op)
```

```python
pd.DataFrame({"predicted OP":prediction_op, "Actual op":test_Y})
```

```python
## see the decision tree
tree.export_graphviz(model, out_file='decision_tree.dot', feature_names=relevant_columns)
# paste the binary file O/P in http://www.webgraphviz.com/
```

## Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier(n_estimators=100)
# train the model
model.fit(train_X, train_Y)

#testing
predicted_op = model.predict(test_X)
metrics.accuracy_score(test_Y, predicted_op)
```
